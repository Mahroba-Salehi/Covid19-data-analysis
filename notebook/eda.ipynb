```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# بارگذاری دیتای پاکسازی شده
df = pd.read_csv("../data/owid_covid_cleaned.csv")
df['date'] = pd.to_datetime(df['date'])

# بررسی آمار توصیفی
df.describe()

# بررسی همبستگی
correlation = df[['new_cases','new_deaths','new_vaccinations','new_tests']].corr()
sns.heatmap(correlation, annot=True)
plt.show()

# نمودار پراکندگی موارد روزانه و مرگ‌ها
sns.scatterplot(data=df, x='new_cases', y='new_deaths', hue='continent')
plt.show()
```
