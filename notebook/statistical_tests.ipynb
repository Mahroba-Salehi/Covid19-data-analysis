```python
import pandas as pd
from scipy.stats import pearsonr

df = pd.read_csv("../data/owid_covid_cleaned.csv")

# همبستگی بین واکسیناسیون و مرگ‌ها
vacc_death_corr, p_value = pearsonr(df['new_vaccinations'], df['new_deaths'])
print(f"Correlation between vaccinations and deaths: {vacc_death_corr:.3f}, p-value: {p_value:.3f}")

# همبستگی بین تست‌ها و موارد مثبت
test_cases_corr, p_value2 = pearsonr(df['new_tests'], df['new_cases'])
print(f"Correlation between tests and cases: {test_cases_corr:.3f}, p-value: {p_value2:.3f}")

# مقایسه قبل و بعد واکسیناسیون
df['vacc_started'] = df['total_vaccinations'] > 0
before_vacc = df[df['vacc_started']==False]['new_deaths'].mean()
after_vacc = df[df['vacc_started']==True]['new_deaths'].mean()
print(f"Mean deaths before vaccination: {before_vacc:.2f}, after vaccination: {after_vacc:.2f}")

# مقایسه کشورها با درآمد بالا و پایین
high_income = df[df['gdp_per_capita'] > df['gdp_per_capita'].median()]['new_cases'].mean()
low_income = df[df['gdp_per_capita'] <= df['gdp_per_capita'].median()]['new_cases'].mean()
print(f"Mean daily cases high income countries: {high_income:.2f}")
print(f"Mean daily cases low income countries: {low_income:.2f}")
```
